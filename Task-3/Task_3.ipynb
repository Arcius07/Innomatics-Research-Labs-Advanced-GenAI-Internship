# Problem Statement 1: Employee Performance Bonus Eligibility

```python
employees = {
"Ravi": 92,
"Anita": 88,
"Kiran": 92,
"Suresh": 85
}
# used a varible for maximum
mx=0
lst=[]
# Core Logic: checks scores of each employes and if it is greater than mx or maximim store that instead
for i in employees.values():
  if i>mx:
    mx=i
# Core Logic: retrieved the keys or names of employees whose score is equal to max score and store them in seperate list
for i in employees.keys():
  if employees[i]==mx:
    lst.append(i)

print("Top Performers Eligible for Bonus:",lst, "(Score: ",mx,")" )
```

# Problem Statement 2: Search Query Keyword Analysis

```python
query = input("Enter Your Query: ").lower()
dic={}
final=""
# isalnum is a inbuild python functions which checks i value is all numbers or alphabets not any special characters ex. ),{,[,%,# etc..
for i in query:
  if i.isalnum() or i == " ":
    # joined the words or alphabets and remove the special charaters
    final+=i
# split breaks the words
wrd= final.split()
# Core Logic: check frequency if a new word appears in dic set the value to 1, else set increase the value by 1
for i in wrd:
  if i in dic:
    dic[i]+=1
  else:
    dic[i]=1
result={}
# Retrieve the values greater than 1
for key, value in dic.items():
  if value>1:
    result[key]=value
print(result)
```

# Problem Statement 3: Sensor Data Validation


```python
sensor_readings = [3, 4, 7, 8, 10, 12, 5]

lst=[]
# Core Logic: iterate through entire list and if the current element(represented by "i") is divisible by 2 i.e. is even append or add that to the list(lst)
for i in range(len(sensor_readings)):
    if sensor_readings[i] % 2 == 0:
        lst.append((i, sensor_readings[i]))

print("Valid Sensor Readings (Hour, Value):")
print(lst)
```

# Problem Statement 4: Email Domain Usage Analysis

```python
emails = [
"ravi@gmail.com",
"anita@yahoo.com",
"kiran@gmail.com",
"suresh@gmail.com",
"meena@yahoo.com"
]
lst=[]
'''
Core Logic: .find() is an inbuild function that retrieves the index of the element that we pass.
Here it retrieves the index of "@" and then moves 1 step right.
"+1:" means moves one step right till the end i.e ":"
'''
for i in emails:
  lst.append(i[i.find("@")+1:])
dic={}
# Core logic: Checks for frequency
for i in lst:
  if i in dic:
    dic[i]+=1
  else:
    dic[i]=1
# Core Logic: calculate the percentage using formula
for key,value in dic.items():
    ans = (value * 100) // len(emails)
    print(key + ": " , str(ans) , "%")
```

# Problem Statement 5: Sales Spike Detection

```python
sales = [1200, 1500, 900, 2200, 1400, 3000]
avg=sum(sales)/len(sales)
day=1
for i in sales:
  if i>round(avg+avg*(30/100)):
    print("Day ",day,": ",i)
  day+=1
```
