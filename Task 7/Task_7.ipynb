# Problem 1: Smart Light Controller

```python
class smart_light:
  def __init__(self, name):
    self.name = name
    self.is_on = False

  def on(self):
    self.is_on = True

  def off(self):
    self.is_on = False

  def status(self):
    if self.is_on:
      print(self.name, " is ON")
    else:
      print(self.name, " is OFF")

name = "Bedroom Light"
mode = "ON"

light = smart_light(name)

if mode == "ON":
  light.on()
elif mode == "OFF":
  light.off()

light.status()
```

# Problem 2: Employee ID Card System

```python
class Employee:
    def __init__(self, name, id, dep):
        self.name = name
        self.id = id
        self.dep = dep

    def id_card(self):
        print("Employee ID Card")
        print("Name: ",self.name)
        print("ID: ",self.id)
        print("Department: ",self.dep)


name = "Rahul"
id = "EMP102"
dep = "IT"

employee = Employee(name, id, dep)
employee.id_card()
```

# Problem 3: Mobile Contact Record

```python
class contact:
    def __init__(self, name, num):
        self.name = name
        self.num = num

    def contact(self):
        print("Contact Saved")
        print("Name: ",self.name)
        print("Phone: ",self.num)


name = "Anita"
phone = "9876543210"

contact = contact(name, phone)
contact.contact()
```

# Problem 4: Product Price Tag Generator


```python
class product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def price_tag(self):
        print("Product: ",self.name)
        print("Price: ₹",self.price)

name = "Headphones"
price = 2499

product = product(name, price)
product.price_tag()
```

# Problem 5: Movie Rating Display System

```python
class movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating

    def detail(self):
        print("Movie: ",self.name)
        print("Rating: ",self.rating," / 5")

name = "Inception"
rating = 4.8

movie = movie(name, rating)
movie.detail()
```

# Problem 6: Delivery Address Manager


```python
class delivery:
    def __init__(self, name, address):
        self.name = name
        self.address = address

    def details(self):
        print("Delivery Details")
        print("Customer: ",self.name)
        print("Address: ",self.address)

name = "Suman"
address = "Hyderabad"

delivery = delivery(name, address)
delivery.details()
```
