# Problem 1: Mobile Recharge Validation System

```python
def validation(x):
  lst=[199, 299, 399, 599]
  if x < 50:
    print("Recharge amount must be ≥ ₹50")
    return False
  if x not in lst:
    print("Invalid Plan")
    return False
  return True
while True:
  x= int(input("Enter Amount: "))
  if validation(x):
    print("Recharge successful")
    break
  else:
    print("Try Again \n")
```

# Problem 2: Inventory Reorder Alert System

```python
def Inventory(dic):
  for key,value in dic.items():
    if value<15:
      print("Reorder Alert")
    else:
      print("Stock OK")
dic={
    "Pen":20,
    "Books":8,
    "Eraser":25,
    "Marker":9,
    "Pencil":20,
    "Sharpener": 14,
    "Ruler": 22,
}
Inventory(dic)
```

# Problem 3: Student Result Processing System

```python
def student_result(lst):
  avg=sum(lst)/len(lst)
  if avg>=50:
    result="Pass"
  else:
    result="Fail"
  return avg,result
marks = [40, 60, 55, 50, 70]
avg,result=student_result(marks)
print("Average Marks: ",avg)
print("Result: ",result)
```

# Problem 4: Cab Fare Estimator with Retry Option

```python
def calculate_fare(distance, peak_hour):
    base_fare=50
    per_km_rate=12
    # calculate normal fare
    fare=base_fare+(distance*per_km_rate)
    # add peak hour extra
    if peak_hour.lower()=="yes":
        fare=fare+(fare*0.25)

    return fare
# retry system using while loop
while True:
    try:
        distance=float(input("Enter distance in km: "))
        if distance<=0:
            print("Distance must be greater than 0.\n")
            continue
        peak_hour=input("Is it peak hour? (yes/no): ")
        fare=calculate_fare(distance,peak_hour)
        print(f"Total Fare: ₹{fare:.2f}")
        break
    except ValueError:
        print("Please enter valid input.\n")
```

# Problem 5: Employee Attendance Eligibility Checker

```python
attendance_list = ["P","A","P","P","A","P","P","P"]
def check_attendance(attendance):
    present_count=0
    # loop to count present days
    for day in attendance:
        if day=="P":
            present_count+=1

    # calculate percentage
    percentage=(present_count/len(attendance))*100
    print(f"Attendance Percentage: {percentage:.2f}%")
    if percentage>=75:
        return "Eligible"
    else:
        return "Not Eligible"
# Calling the function and saving the result in result variable
result = check_attendance(attendance_list)
print("Status:", result)
```

# Problem 6: Password Strength Checker


```python
def check_password_strength(password):
    digit =False
    special =False
    special_chars ="@#$"
    # check minimum length of the password
    if len(password)<8:
        return "Weak Password (Minimum length is 8)"

    # loop over string to check strength
    for char in password:
        if char.isdigit():
            digit=True
        if char in special_chars:
            special=True

    # check conditions
    if digit and special:
        return "Strong Password"
    else:
        return "Weak Password (Must contain digit and special character)"
password=input("Enter password: ")
result=check_password_strength(password)
print(result)
```
