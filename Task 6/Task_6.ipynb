# Problem Statement 1:Smart Parking Lot Management System

```python
def parking(Capacity,logs):
  count=0
  peak=0
  for i in logs:
    if i=="IN":
      count+=1
    elif i=="OUT":
      if count>0:
        count-=1
    peak=max(peak,count)
  print("Currently Parked Vehicles: ",count)
  print("Peak Parking Usage: ",peak)

  if count>=Capacity:
    print("Parking Status: ALERT! Exceeds Capacity")
  else:
    print("Parking Status: Available")

capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]
parking(capacity,logs)
```

# Problem Statement 2:Online Food Delivery Time Estimator

```python
def estimate(distance,traffic,weather):
  time = (distance / 30) * 60
  if traffic =="Low":
    time +=5
  elif traffic =="Medium":
    time +=10
  elif traffic =="High":
    time +=25

  if weather =="Cloudy":
    time +=5
  elif weather =="Rainy":
    time +=15
  print("Estimated Delivery Time:", time, "minutes")

distance=8
traffic="High"
weather="Rainy"
estimate(distance, traffic, weather)
```

# Problem Statement 3:Movie Theatre Seat Occupancy Analyzer

```python
def occupancy_analyze(total,booked):
  booked= len(booked)
  occupy = (booked/total)*100

  if occupy ==100:
    status= "HouseFull"
  elif occupy>=75:
    status ="Almost Full"
  else:
    status ="Seats Available"

  if occupy>=90:
    suggestion= "Suggestion: Open additional shows"
  else:
    suggestion= "Suggestion: No additional shows needed"

  print("Occupancy:", int(occupy), "%")
  print("Show Status:", status)
  print(suggestion)

total =200
seats =[1]*150
occupancy_analyze(total, seats)
```

# Problem Statement 4:Cloud Server Load Classification System

```python
def server_load(readings):
  total=0
  for i in readings:
    total+=i
  avg= total/len(readings)
  if avg<50:
    status="Normal"
  elif avg>=50 and avg<=80:
    status="Warning"
  else:
    status="Critical"
  print("Average CPU Load:", int(avg),"%")
  print("Server Status:", status)

readings = [45, 60, 70, 85, 90]
server_load(readings)
```

#  Problem Statement 5:Smart Classroom Resource Usage Monitor


```python
def resource_usage(usage):
  overuse=[]
  for resoruce ,time in usage.items():
    if time>8:
      overuse.append(resoruce)
  if overuse:
    print("Overused Resource: ",",".join(overuse))
    print("Energy Alert: Yes")
  else:
    print("No Overused Resources")
    print("Energy Alert: No")

usage = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4}
resource_usage(usage)
```

#  Problem Statement 6: Online Event Registration Capacity Controller

```python
def registration(capacity,reg):
  confirmed=0
  waitlist= 0
  for i in range(reg):
    if confirmed < capacity:
        confirmed += 1
    else:
        waitlist += 1

  if confirmed==capacity:
      status= "Closed"
  else:
      status= "Open"

  print("Confirmed Registrations: ",confirmed)
  print("Waitlist Registrations: ",waitlist)
  print("Registration Status: ",status)

capacity =100
reg=105
registration(capacity, reg)
```
